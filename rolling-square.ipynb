```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, ElasticNet
import csv
import seaborn as sns
from typing import Any
import math
```

```python
length = 4
x = np.arange(0, length, 0.0001)
len(x)
```

```python
p = [x[i] - ((np.pi)/(2))*math.floor(((x[i])/(((np.pi)/(2))))) for i in range(0, len(x))]
```

```python
h = [x[i] - ((np.pi)/(4))*math.floor(((x[i])/(((np.pi)/(4))))) for i in range(0, len(x))]
```

```python
y = np.zeros(len(x))
y_surface = np.zeros(len(x))
```

```python
for xval in range(len(x)):
    if p[xval] < np.pi/4:
        y[xval] = math.sqrt(2) - (1/math.cos(h[xval]/math.sqrt(2)))
    else:
        y[xval] = math.sqrt(2) - (1/math.cos(((math.pi/4) - h[xval])/math.sqrt(2)))

for xval in range(len(x)):
    y_surface[xval] = y[xval] - y[int(math.pi/4)]
```

```python
y_surface 
```

```python
plt.plot(x, y_surface)
plt.show()
```
